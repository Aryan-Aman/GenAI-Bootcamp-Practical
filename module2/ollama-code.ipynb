```python
import warnings

warnings.filterwarnings("ignore")
```

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
local_llm = ChatOllama(model="deepseek-r1:latest", temperature=0.9)
```

```python
local_llm_cloud = ChatOllama(model="nemotron-3-super:cloud", temperature=0.9)
```

```python
messages = [
    SystemMessage(content="You are a helpful AI assistant."),
    HumanMessage(content="Explain what Ollama is in simple terms.")
]
```

```python
resp = local_llm.invoke(messages)
print(resp.content)
```

```python
response2 = local_llm_cloud.invoke(messages)
print(response2.content)
```
