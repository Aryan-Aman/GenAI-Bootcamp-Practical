```python
from huggingface_hub import HfApi
```

```python
api = HfApi()
```

**Detail of any Repo**

```python
model_info = api.model_info(
    repo_id="google-bert/bert-large-uncased-whole-word-masking"
)
for key, value in model_info.__dict__.items():
    print(f"\n{key}:")
    print(value)
```

**Search about any models id/downloads**

```python
models = api.list_models(search='llama', limit = 20)
for model in models:
  print (model.modelId, "->", model.downloads)
```

**Access any model from HF**

```python
from huggingface_hub import InferenceClient
```

```python
client = InferenceClient(model = "meta-llama/Llama-3.1-8B-Instruct")
```

```python
response  = client.chat.completions.create(
    messages=[
        {"role": "user", "content":"Explain Hugging Face in 50 words"}
    ], max_tokens=300
)
print(response)
```

```python
from langchain_huggingface import ChatHuggingFace
from langchain_huggingface import HuggingFaceEndpoint
```

```python
llm = HuggingFaceEndpoint(
    repo_id = "meta-llama/Llama-3.1-8B-Instruct",
    task = "text-generation",
    max_new_tokens= 500,
    do_sample = True,
    temperature = 0.1,
    streaming = True
)
```

```python
chat_model =ChatHuggingFace(llm = llm)
```

```python
chat_model.invoke("hello who are humans?")
```
