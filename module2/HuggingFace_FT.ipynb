**AutoTokenizer flow**

```python
from transformers import AutoTokenizer
```

```python
dataset = load_dataset(
    "stanfordnlp/imdb",
    split="train"
)
```

```python
print("Before Tokenization:")
print(dataset[0])
```

```python
# Load tokenizer for a specific pretrained model
# This downloads the tokenizer configuration and vocabulary
# Tokenizer must match the model you will train/use
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)
```

```python
def tokenize_function(batch):

    # batch["text"] contains raw sentences/reviews
    # tokenizer converts text -> input_ids and attention_mask
    tokens = tokenizer(
        batch["text"],

        # Cut text if it exceeds max length
        truncation=True,

        # Add padding so all samples have same length
        padding="max_length",

        # Maximum number of tokens per sample
        max_length=128
    )

    return tokens
```

```python
# Apply tokenizer to complete dataset
# batched=True sends multiple examples at once for faster processing
tokenized_dataset = dataset.map(
    tokenize_function,
    batched=True
)
```

```python
print("\nAfter Tokenization:")
print(tokenized_dataset[0])
```

```python
text = "I love Gen AI"
```

```python
otpt = tokenizer(text)
```

```python
print(otpt)
```

```python
tokens = tokenizer.convert_ids_to_tokens(otpt["input_ids"])
print(tokens)
```

**Custom tokenizer training pipeline**
*Raw data ->clean text ->train tokenizer ->save ->convert to HFtokenizer ->save*

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers, decoders
```

```python
#create training data
training_data = [
    "I love machine learning",
    "I love artificial intelligence",
    "AI is powerful"
]
```

```python
#create tokenizer with BPE algorithm
from tokenizers.models import BPE
```

```python
tokenizer = Tokenizer(
    BPE(unk_token="[UNK]")
)
```

```python
# Define how text is split
from tokenizers.pre_tokenizers import Whitespace
tokenizer.pre_tokenizer = Whitespace()
```

```python
# Define training rules

from tokenizers.trainers import BpeTrainer

trainer = BpeTrainer(
    vocab_size=100,
    special_tokens=[
        "[PAD]",
        "[UNK]"
    ]
)
```

```python
#Train tokenizer

tokenizer.train_from_iterator(
    training_data,
    trainer=trainer
)
```

```python
# Test tokenizer

output = tokenizer.encode(
    "I love machine learning"
)

print(output.tokens)   # tokens
print(output.ids)      # token IDs
```

```python
# Save tokenizer

tokenizer.save(
    "custom_tokenizer.json"
)
```

```python
# Load tokenizer later

new_tokenizer = Tokenizer.from_file(
    "custom_tokenizer.json"
)

print(
    new_tokenizer.encode("AI is powerful").tokens
)
```

```python
#save custom tokenizer in a Hugging Face compatible folder
from transformers import PreTrainedTokenizerFast
```

```python
tokenizer = Tokenizer.from_file(
    "custom_tokenizer.json"
)
```

```python
# Convert to HuggingFace tokenizer format
hf_tokenizer = PreTrainedTokenizerFast(
    tokenizer_object=tokenizer,

    # special tokens
    unk_token="[UNK]",
    pad_token="[PAD]"
)
```

```python
# Save as HuggingFace folder
hf_tokenizer.save_pretrained(
    "./my_custom_tokenizer"
)
```

**Load Models**

```python
#load tokenizer
from transformers import AutoTokenizer


tokenizer = AutoTokenizer.from_pretrained(
    "./my_custom_tokenizer"
)
```

```python
#load pretrained model
from transformers import AutoModel


model = AutoModel.from_pretrained(
    "bert-base-uncased"
)
```

*Resize model embeddings for tokenizer*
my custom tokenizer has diff vocab
eg here 64 but bert 30522

```python
print(len(tokenizer))
```

```python
#update embeddings
model.resize_token_embeddings(
    len(tokenizer)
)
#Now model understand tokenizer ids
```

```python
#tokenize input
text = "I love machine learning"


inputs = tokenizer(
    text,
    return_tensors="pt"
)


print(inputs)
```

```python
#pass into model
outputs = model(
    **inputs
)


print(outputs.last_hidden_state.shape)
```
