# Classical Word Embedding

**Goal:** load a pretrained Word2Vec model and inspect word vectors, similarity, analogy behavior, and a simple sentence embedding.

**Flow:** Load pretrained model → query words → compare semantic similarity → run analogy arithmetic → build sentence vector from word averages

```python
from gensim.models import Word2Vec, KeyedVectors
import numpy as np
import gensim.downloader as api
import warnings
```

## Step 1: Load pretrained embeddings

- `gensim.downloader` fetches the pretrained Google News Word2Vec model.
- This avoids training from scratch and gives immediate access to 300-dimensional word vectors.

```python
model=api.load("word2vec-google-news-300")
```

## Step 2: Inspect word lookup behavior

- Retrieve vectors for words that exist in the vocabulary.
- Check vector dimension.
- Try a phrase lookup to observe why pretrained Word2Vec is word-level, not sentence-level.

```python
print(model["aman"])
len(model["aman"])
```

```python
try:
    model["i am aman"]  # will fail because this pretrained model is word-level
except KeyError:
    pass
```

```python
model["men"]
```

## Step 3: Run semantic queries

- `most_similar()` returns nearest neighbors.
- `similarity()` compares closeness between two words.
- `doesnt_match()` finds the semantic outlier in a list.

```python
print(model.most_similar("aryan"))
print(model.most_similar("america"))
```

```python
model.similarity("mango", "vegetable")
model.similarity("mango", "fruit")
model.similarity("mango", "car")
```

```python
model.doesnt_match(["breakfast", "lunch", "dinner", "car"])
model.doesnt_match(["king", "girl", "man", "pawn", "god"])
```

## Step 4: Analogy with vector arithmetic

- Combine word vectors using subtraction and addition.
- Example pattern: `king - man + woman`.
- The nearest result shows how relationships are encoded in vector space.

```python
vec=model['king'] - model['man'] + model['woman']
```

```python
model.most_similar([vec])
```

## Step 5: Build a simple sentence embedding

- Split a sentence into words.
- Keep only words present in the vocabulary.
- Collect their vectors.
- Average them with NumPy to get one sentence-level vector.

```python
sentenc = 'I am aman and i am a gen ai engineer'
```

```python
words=sentenc.lower().split()
```

```python
word_vectors=[]
```

```python
for word in words:
    if word in model:
        print(word)
        word_vectors.append(model[word])
    else:
        print(f"'{word}' not in vocabulary, skipping")
```

```python
word_vectors
```

```python
sentenc_vec =np.mean(word_vectors, axis=0) #axis=0 means columsn basis
sentenc_vec
```
